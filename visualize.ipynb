```python
from matplotlib import pyplot as plt
import pandas as pd
```

```python
# Plot a csv file with two columns
df = pd.read_csv("data/random_walk.csv")
df2 = pd.read_csv("data/wiener_process.csv")
df3 = pd.read_csv("data/gbm.csv")
df4 = pd.read_csv("data/jump_diff.csv")
plt.plot(df.iloc[:, 0], df.iloc[:, 1])  # First column as x, second column as y
plt.xlabel('Index')
plt.ylabel('Values')
plt.title('Random Walk')
plt.show()

plt.plot(df2.iloc[:, 0], df2.iloc[:, 1])  # First column as x, second column as y
plt.xlabel('Index')
plt.ylabel('Values')
plt.title('Wiener Process')
plt.show()

plt.plot(df3.iloc[:, 0], df3.iloc[:, 1])  # First column as x, second column as y
plt.xlabel('Index')
plt.ylabel('Values')
plt.title('Geometric Brownian Motion')
plt.show()

plt.plot(df4.iloc[:, 0], df4.iloc[:, 1])  # First column as x, second column as y
plt.xlabel('Index')
plt.ylabel('Values')
plt.title('Jump Diffusion')
plt.show()  
```
