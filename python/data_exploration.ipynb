```python
import yfinance as yf
import datetime as dt
import numpy as np
import pandas as pd
```

```python
start_date = dt.datetime(1900, 1, 1)
end_date = dt.datetime.today()

gold = yf.download("GC=F", start=start_date, end=end_date)
```

### Create daily log returns for gold prices to analyze volatility and trends over time

```python
daily_log_returns = np.log(gold.Close).diff().dropna()
daily_log_returns
```

### Find historic events that caused significant price movements in gold and analyze their impact on the market

- 2008 financial crisis june 2007 - december 2008
- COVID-19 pandemic december 2019- june 2020
- liberation day march 2025 - june 2025 
- 9/11 2001 couple of months after the attack
- price hike 2026

Maybe change the the intervals in a way such that they are more comparable in terms of the number of trading days. For example, you could analyze the 6 months following each event to see how gold prices reacted in a similar time frame. This would allow for a more consistent comparison of the market's response to different events.

```python
df_times = pd.concat(
    {
        "All Time return": daily_log_returns,
        "Financial Crisis": daily_log_returns.loc["2007-06-01":"2008-12-31"],
        "COVID-19 Pandemic": daily_log_returns.loc["2019-12-01":"2020-06-30"],
        "Liberation Day": daily_log_returns.loc["2025-03-01":"2025-06-30"],
        "9/11": daily_log_returns.loc["2001-09-11":"2001-11-30"],
        "2026 Rollercoaster": daily_log_returns.loc["2026-01-01":"2026-02-06"],
    },
    axis=1
)
```

```python
volas = df_times.apply(lambda x: x.std() * np.sqrt(252)) # annualize volatility by multiplying the standard deviation of daily returns by the square root of 252 trading days in a year
volas.sort_values(ascending=False)
```
